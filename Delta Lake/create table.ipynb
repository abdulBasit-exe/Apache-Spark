```python
from delta.tables import *
```

```python
DeltaTable.createOrReplace(spark) \
    .tableName("delta_table_demo") \
    .addColumn("emp_id", "INT") \
    .addColumn("emp_name", "STRING") \
    .addColumn("gender", "STRING") \
    .addColumn("salary", "INT") \
    .addColumn("department", "STRING") \
    .property("description", "demo table created for learning").execute()
    # .location("/FileStore/tables/delta/demo_table") \
```

```python
%sql
select * from delta_table_demo;
```

### insert data into delta table ###

```python
spark.sql("""
    INSERT INTO delta_table_demo (emp_id, emp_name, gender, salary, department)
    VALUES 
    (1, 'Ali', 'Male', 50000, 'Engineering'),
    (2, 'Aisha', 'Female', 55000, 'Finance'),
    (3, 'Zaid', 'Male', 48000, 'HR')
""")
```
