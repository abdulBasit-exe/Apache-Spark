```python
dbutils.widgets.text("username", "Basit", "Enter your name")
name = dbutils.widgets.get("username")
print(f"Hello, {name}!")
```

```python
dbutils.widgets.dropdown("env", "dev", ["dev", "test", "prod"], "Select Environment")
env = dbutils.widgets.get("env")
print(f"Selected Environment: {env}")
```

```python
dbutils.widgets.combobox("country", "Pakistan", ["Pakistan", "India", "Bangladesh"], "Select or enter a country")
country = dbutils.widgets.get("country")
print(f"Country: {country}")
```

```python
dbutils.widgets.multiselect("technologies", "Spark", ["Spark", "Hadoop", "Kafka", "Airflow"], "Select Technologies")
selected = dbutils.widgets.get("technologies")
print(f"Selected Technologies: {selected}")
```
