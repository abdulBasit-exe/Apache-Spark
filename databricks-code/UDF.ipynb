### UDF

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Define schema
schema = StructType([
    StructField("UserID", StringType(), True),
    StructField("Username", StringType(), True),
    StructField("Age", IntegerType(), True),
    StructField("Country", StringType(), True)
])

# Sample data
data = [
    ("U001", "basit", 23, "Pakistan"),
    ("U002", "sumair123", 25, "India"),
    ("U003", "uzair.khan", 29, "Bangladesh"),
    ("U004", "hamza", 22, "Pakistan"),
    ("U005", "zoya", 28, "India"),
    ("U006", "ahmed.shah", 26, "Nepal")
]

# Create DataFrame
df = spark.createDataFrame(data, schema)
df.show()
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

# Define UDF
def add_email_domain(username):
    return f"{username}@gmail.com" if username else None

# Register UDF
add_email_udf = udf(add_email_domain, StringType())

# Apply UDF
df_with_email = df.withColumn("Email", add_email_udf(df["Username"]))

# Show result
df_with_email.show(truncate=False)
```

```python
from pyspark.sql.types import StructType, StructField, LongType, StringType, IntegerType, DateType
from datetime import date

# Define schema
schema = StructType([
    StructField("employee_id", LongType(), True),
    StructField("Name", StringType(), True),
    StructField("doj", DateType(), True),
    StructField("employee_dept_id", LongType(), True),
    StructField("salary", IntegerType(), True)
])

# Sample data
data = [
    (10, "Michael Robinson", date(1999, 6, 1), 100, 2000),
    (20, "James Wood", date(2003, 3, 1), 200, 8000),
    (30, "Chris Andrews", date(2005, 4, 1), 100, 6000),
    (40, "Mark Bond", date(2008, 10, 1), 100, 7000),
    (50, "Steve Watson", date(1996, 2, 1), 400, 1000),
    (60, "Mathews Simon", date(1998, 11, 1), 500, 5000),
    (70, "Peter Paul", date(2011, 4, 1), 600, 5000)
]

# Create DataFrame
empDF = spark.createDataFrame(data, schema=schema)

# Show DataFrame
empDF.show(truncate=False)
```

```python
import pyspark.sql.functions as f

def rename_columns(rename_df):

    for column in rename_df.columns:
        new_coulmn = 'Col_' +column
        rename_df = rename_df.withColumnRenamed(column, new_coulmn)

    return rename_df
```

```python
renamed_df = rename_columns(empDF)
display(renamed_df)
```

### UDF to convert name in upper case

```python
from pyspark.sql.functions import col, upper


def upperCase_col(df):

  empDF_upper = df.withColumn('name_upper', upper(df.Name))

  return empDF_upper
```

```python
up_case_df = upperCase_col(empDF)

display(up_case_df)
```
