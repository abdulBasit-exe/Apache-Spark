```python
data1 = [("Basit", "Python", 90),
         ("Ali", "SQL", 85)]

df1 = spark.createDataFrame(data1, ["name", "skill", "score"])
df1.show()
```

```python
# Second DataFrame
data2 = [("Basit", "Python", 90),  # Duplicate row
         ("Sumair", "Spark", 88)]

df2 = spark.createDataFrame(data2, ["name", "skill", "score"])
df2.show()
```

```python
df1.union(df2).show()
```

```python
df1.union(df2).distinct().show()
```

```python
df1.unionAll(df2).show()
```

```python
data3 = [("Muet", "Basit", "Python", 90),  # Duplicate row
         ("Muet","Sumair", "Spark", 88)]

df3 = spark.createDataFrame(data3, ["UniName","name", "skill", "score"])
df3.show()
```

```python
df1.unionByName(df3, allowMissingColumns=True).show()
```
