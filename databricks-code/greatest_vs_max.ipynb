```python
import pyspark.sql.functions as fn 

data = [
    ("Ali", 85, 90, 78),
    ("Zainab", 76, 88, 92),
    ("Omar", 55, 65, 60),
    ("Fatima", 95, 91, 89),
    ("Hamza", 40, 45, 50)
]

columns = ["Name", "Math", "Science", "English"]

df = spark.createDataFrame(data, columns)

# Show original data
df.display()
```

### Greatest ###

```python
# df.select("*", fn.greatest(df.Math, df.Science, df.English).alias("Highest")).display()

great_df = df.withColumn("Greatest", fn.greatest(df.Math, df.Science, df.English))
great_df.display()
```

## Lowest ##

```python
df.select("*", fn.least(df.Math, df.Science, df.English).alias("Lowest")).show()
```

## Max ##

```python
df.select(fn.max("Math"), fn.max("Science"), fn.max("English")).show()
df.select(fn.min("Math"), fn.min("Science"), fn.min("English")).show()
```
