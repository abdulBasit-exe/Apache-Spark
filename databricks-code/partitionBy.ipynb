```python
sales_df = spark.read.csv("dbfs:/FileStore/Sales_2017/Sales_2017.csv", header=True, inferSchema=True)
sales_df.show(10)
```

```python
sales_df.count()
```

```python
# Number of partitions per DataFrame
print("sales_df: ", sales_df.rdd.getNumPartitions()) 
```

```python
# Number of records per partiton
from pyspark.sql.functions import spark_partition_id

sales_df.withColumn("partitionID", spark_partition_id()).groupBy("partitionID").count().show()
```

```python
sales_df_partitioned = sales_df.select(sales_df.OrderDate, "CustomerKey", "ProductKey").repartition(5)
```

```python
print(sales_df_partitioned.rdd.getNumPartitions())
```
