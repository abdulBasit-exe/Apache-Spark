```python
array_skills = [
    ("Basit", ['Python', 'SQL', 'Azure']),
    ("Sumair", ['Python', 'RAG', 'LLM']),
    ("Uzair", ['Java', 'AWS', None])
]

df_skills = spark.createDataFrame(data=array_skills, schema=['name', 'skils'])
df_skills.printSchema()
display(df_skills)
```

```python
map_level = [
    ("Basit", {"Python": "Advanced", "SQL": "Intermediate", "Azure": "Beginner"}),
    ("Sumair", {"Python": "Advanced", "RAG": "Intermediate", "LLM": "Beginner"}),
    ("Uzair", {"Java": "Advanced", "AWS": "Intermediate", "C++": None})
]
```

```python
df_level = spark.createDataFrame(data=map_level, schema=['name', 'level'])
df_level.printSchema()
display(df_level)
```

### Explode Array field

```python
from pyspark.sql.functions import explode

df2 = df_skills.select(df_skills.name, explode(df_skills.skils))

df2.printSchema()
display(df2)
```

### Explode Map Field

```python
from pyspark.sql.functions import explode

df3 = df_level.select(df_level.name, explode(df_level.level))

df3.printSchema()
display(df3)
```

```python
from pyspark.sql.functions import explode_outer
```

```python
display(df_skills.select(df_skills.name, explode_outer(df_skills.skils)))
display(df2)
```

```python
display(df_level.select(df_level.name, explode_outer(df_level.level)))
display(df3)
```
