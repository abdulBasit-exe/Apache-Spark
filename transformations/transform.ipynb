```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.csv(path="/home/abdul-basit/Documents/Data_Engineering/Spark/Chapter-6/Titanic.csv", inferSchema=True, header=True)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col
```

```python
df.select(col("Pclass").alias("PassengerClass"), "Survived", "Fare").show()
```

### Filter ### 

```python
df = df.filter(col("Pclass")>2)
```

### Rename column ###

```python
df = df.withColumnRenamed("Pclass", "PassengerClass")
```
